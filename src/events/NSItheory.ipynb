```python
eem_range = np.linspace(-0.3, 0.3, 29)
eet_range = np.linspace(-0.3, 0.3, 29)
E_range = [5, 25, 200]
param_list = []
for eet in eet_range:
    for eem in eem_range:
        dicta = dc_params_nsi.copy()
        dicta.update({"e_em": eem, "e_et": eet})
        param_list.append(dicta)
    
Pme = []
Pamae = []
Pme_SI = P_num_over_E("m", E_range, nsi=False, params=dc_params_nsi)[0, :]
Pamae_SI = P_num_over_E("m", E_range, nsi=False, anti=True,params=dc_params_nsi)[0, :]
for p in param_list:
    prob = P_num_over_E("m", E_range, nsi=True, params=p)[0, :] - Pme_SI
    Pme.append(prob)
    prob = P_num_over_E("e", E_range, nsi=True, anti=True, params=p)[0, :] - Pamae_SI
    Pamae.append(prob)
Pme = np.array(Pme).reshape(len(eem_range), len(eet_range), len(E_range))
Pamae = np.array(Pamae).reshape(len(eem_range), len(eet_range), len(E_range))
```

```python
fig, ax = plt.subplots(2, 3, figsize=(12, 8), sharey=True)
titles = ["5 GeV", "25 GeV", "200 GeV"]

# cbar
color_ranges = [[0.0, 0.5], [0.8, 1]]
cbar_height = 0.335
cbar_width = 0.02
cbar_left = 0.935
cbar_labels = [r'$\nu_\mu \to \nu_e$', r'$\bar{\nu}_\mu \to \bar{\nu}_e$']
cbar_positions = [[cbar_left, 0.507, cbar_width, cbar_height], [cbar_left, 0.105, cbar_width, cbar_height]] # [left, bottom, width, height]
cbar_colors = ["Reds", "Blues"]

for i, data_array in enumerate([Pme, Pamae]):
    for j in range(len(E_range)):
        data = data_array[:, :, j].T
        data = np.abs(data)
        im = ax[i][j].pcolormesh(data, cmap=cbar_colors[i], vmin=color_ranges[i][0], vmax=color_ranges[i][1])
        ax[i][j].tick_params(
            axis="both",
            direction="in",
            which="both",
            labelsize=15,
            width=1,
            length=4,
            zorder=20,
        )
        ax[0][j].set_title(titles[j], fontsize=20)

        # Set the axis limits in terms of array indices
        ax[i][j].set_xlim(0, len(eem_range)-1)
        ax[i][j].set_ylim(0, len(eet_range)-1)

        # Create tick positions and labels
        xtick_positions = np.linspace(0, len(eem_range)-1, 5)
        ytick_positions = np.linspace(0, len(eet_range)-1, 5)
        xtick_labels = np.linspace(-0.3, 0.3, 5)
        ytick_labels = np.linspace(-0.3, 0.3, 5)

        ax[i][j].set_xticks(xtick_positions)
        ax[i][j].set_yticks(ytick_positions)
        
        ax[i][j].set_xticklabels([f'{x:.2f}' for x in xtick_labels])
        ax[i][j].set_yticklabels([f'{y:.2f}' for y in ytick_labels])
        
        # Adjust label positions to prevent overlap
        ax[i][j].tick_params(axis='x', pad=8)  # Add padding between x-axis and its labels
        ax[i][j].tick_params(axis='y', pad=3)  # Add padding between x-axis and its labels


        ax[1][j].set_xlabel(r'$\epsilon_{e\mu}$', fontsize=25)
        ax[i][0].set_ylabel(r'$\epsilon_{e\tau}$', fontsize=25)
    
    # Add colorbar for each row
    cbar_ax = fig.add_axes(cbar_positions[i])  
    cbar = fig.colorbar(im, cax=cbar_ax)
    cbar.ax.tick_params(labelsize=15)
    cbar.set_label(cbar_labels[i], fontsize=25)




plt.suptitle(r'$|{P_{\mu e}^{NSI} - P_{\mu e}^{SI}} |$', fontsize=35)
plt.tight_layout()
plt.subplots_adjust(right=0.9)

#save(fig, "eem_eet_prob.pdf")
```
